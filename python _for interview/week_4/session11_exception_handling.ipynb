There are 2 stages where error may happen in a program

- During compilation -> Syntax Error
- During execution -> Exceptions

### Syntax Error

- Something in the program is not written according to the program grammar.
- Error is raised by the interpreter/compiler
- You can solve it by rectifying the program


In Python, errors can be broadly categorized into three main types:

1. Syntax Errors: These occur when the code violates the grammatical rules of the Python language. The Python interpreter cannot even parse the code and will raise a SyntaxError, preventing the program from running. Examples include forgetting a colon, mismatched parentheses, or incorrect indentation (which specifically raises an IndentationError).

""  ********************************************************************************************************************"

2. Runtime Errors (Exceptions): These errors occur while the program is executing, even if the syntax is correct. When a runtime error occurs, an "exception" is raised, which can be handled using try-except blocks. If an unhandled exception occurs, the program will terminate. 

Common examples include:

- NameError: Using an undefined variable.

- TypeError: Performing an operation on incompatible data types.

- IndexError: Accessing an index outside the bounds of a list or other sequence.

- KeyError: Accessing a non-existent key in a dictionary.

- ZeroDivisionError: Attempting to divide by zero.

Logical Errors: These are the most subtle type of error, as they do not cause the program to crash or raise an exception. Instead, the program runs but produces incorrect or unexpected results due to flaws in the program's logic. Identifying and fixing logical errors often requires careful debugging and testing.

```python
# try catch demo
try:
  with open('sample4.txt','r') as f:
    print(f.read())
except:
  print('sorry file not found')
```

```python
# catching specific exception
try:
  m=5
  f = open('sample1.txt','r')
  print(f.read())
  print(m)
  print(5/2)
  L = [1,2,3]
  L[100]
except FileNotFoundError:
  print('file not found')
except NameError:
  print('variable not defined')
except ZeroDivisionError:
  print("can't divide by 0")
except IndexError:
  print("index out of range")
except Exception as e:
  print(e)
```

```python
# else
try:
  f = open('sample1.txt','r')
except FileNotFoundError:
  print('file nai mili')
except Exception:
  print('kuch to lafda hai')
else:
  print(f.read())

```

```python
# finally
# else
try:
  f = open('sample1.txt','r')
except FileNotFoundError:
  print('file nai mili')
except Exception:
  print('kuch to lafda hai')
else:
  print(f.read())
finally:
  print('ye to print hoga hi')
```

```python
class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('bhai teri to lag gayi')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')



obj = Google('nitish','nitish@gmail.com','1234','android')

try:
  obj.login('nitish@gmail.com','1234','window')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('database connection closed')

```
