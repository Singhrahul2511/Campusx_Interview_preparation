### Super Keyword

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

```python
# using super outside the class
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")
    # price private attrivute ka ye generator hai
    def get_price(self):
        return self.__price

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        super().buy()

s=SmartPhone(20000, "Apple", 13)

s.buy()
s.get_price()
```

```python
# can super access parent ka data?
# using super outside the class
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        print ("Buying a smartphone")
        # syntax to call parent ka buy method
        print(super().brand) # supper() se attrivute ko access nhi kr skte ho

s=SmartPhone(20000, "Apple", 13)

s.buy()

# supper use always inside of class
# variable,attribute ko access nhi kr skte ho
# always used inside child class
```

```python
# super -> constuctor
# super()-> class ke andar hi only use hota hai
# yaha ye dekhne ko mila ki (phle ye hota tha ki if constructor is available in child then Parent constructor will not get colled) but yaha pr
# super()--keyword ke help se dono ko call kiya gya aur use me laya gya hai
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        self.brand = brand
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", 12, "Android", 2)

print(s.os)
print(s.brand)

# super ka use krke aap Parent ke method ko use kr pate ho jaise kki yaha parent ke (cunstructor) ko call krke kiye hmlog
```

##### Inheritance in summary

- A class can inherit from another class.

- Inheritance improves code reuse

- Constructor, attributes, methods get inherited to the child class -> only public not private (for access private we need to use Getter method)

- The parent has no access to the child class

- Private properties of parent are not accessible directly in child class

- Child class can override the attributes or methods. This is called method overriding

- super() is an inbuilt function which is used to invoke the parent class methods and constructor

```python
class Parent:

    def __init__(self,num):
      self.__num=num

    def get_num(self):
      return self.__num

class Child(Parent):

    def __init__(self,num,val):
      super().__init__(num)
      self.__val=val

    def get_val(self):
      return self.__val

son=Child(100,200)
print(son.get_num())
print(son.get_val())
```

```python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
# single inheritance bcoz (.buy())--> will access phone class ,method
```

```python
# multilevel
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# Multiple
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def review(self):
        print ("Customer review")
    # def buy(self):
    #     print("buying product") # age is buy methos ko un-hash kr diya jaye to phle yhi call hoga (MRO)-> mrthod resolution order

class SmartPhone(Product,Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product): # yhi methos resolution order(MRO) jiska name phle aaya hai usi ko inherite krenge 
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy() # krne pr jiska name phle likha huaa hai usi ka method call hoga(phone,product)
```

# **************** POLYMORPHISM *******************

- Polymorphism means "many forms". It refers to the ability of an entity (like a function or object) to perform different actions based on the context.

- Technically, in Python, polymorphism allows same method, function or operator to behave differently depending on object it is working with. This makes code more flexible and reusable.

### Polymorphism

- Method Overriding
- Method Overloading
- Operator Overloading

```python
#method overriding -->  

class product:
    def __init__(self):
        pass
    def buy(self):
        print("hhh")
class item(product):
    # def __init__(self): # ye constructor overriding hai bhai
        # pass
    def buy(self):
        print("ttgdd") 
obj = item()
obj.buy()
```

```python
class Shape():
    def area(self,radius):
        return 3.14*radius*radius
    def area(self,l,b):
        return l*b
# method overloading me inside class same name ka function hota hai (lekin unka o/p diff hota hai diff i/p ke sath)
s = Shape()
s.area(2)
s.area(3,4)

#method overloading me aapka code clean dikhai deti hai jyada jldi smjh aata hai
```

```python
#method overloading using default agrgument --b = 0
class Shape:

  def area(self,a,b=0):#python me to method overloading nhi hota hai lekin (b=0)default argument se hm ye implement kr skte hai
    #ye smarter way hai
    if b == 0:
      return 3.14*a*a
    else:
      return a*b

s = Shape()

print(s.area(2))
print(s.area(3,4))

# method overloading do not supported by python but can be implemented using (Default-Argument)
```

### Abstraction

- Abstraction means hiding complex implementation details and showing only the essential features of an object.

- It focuses on what a system does, not how it does it.

- abstration se jo higher level ka class hai wo lower level me class ko kuch implement krne keliye majboor kr skta hai jaise ki BankApp ka security method ko override krna hi hoga MobileApp me

```python
from abc import ABC,abstractmethod
class BankApp(ABC): #abstract class ak aesa abstract class hota hai jisme atleast 1 abstract method hota hai

  def database(self):
    print('connected to database')

  @abstractmethod #ye abstract class me hai to ab isko niche ke claas me define krna hi pdega tbhi run hoga
  def security(self):
    pass
    
  # @abstractmethod
  # def display(self):
  #   pass


#abstration is more about hiding important design and encapsulation is like data security
```

```python
from abc import ABC, abstractmethod
class BankApp(ABC):
    def database(self):
        print("Database connection sucessfully completed")
    @abstractmethod
    def security(self):
        pass

class MobileApp(BankApp):
    def login(self):
        print("login into mobile")
    def security(self):
        return super().security()
    
obj = MobileApp()
obj.security()
```
