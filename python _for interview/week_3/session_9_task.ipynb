### `Problem-4:` Write a program that has an abstract class Polygon. Derive two classes Rectangle and Triangle from Polygon and write methods to get the details of their dimensions and hence calculate the area.

````python
from abc import ABC,abstractmethod

class Polygon(ABC):

    @abstractmethod
    def get_dim(self):
        pass
    @abstractmethod
    def area(self):
        pass

class Rectangle(Polygon):
    def get_dim(self):
        self.length = float(input("Enter Length of Rectangle: "))
        self.breadth = float(input("Enter Breadth of Rectangle: "))

    def area(self):
        return self.length * self.breadth

class Triangle(Polygon):
    def get_dim(self):
        self.base = float(input("Enter base of Triangle: "))
        self.height = float(input("Enter Height of Triangle: "))


    def area(self):
        return 0.5*self.base*self.height


# this is completely for Rectangle
r = Rectangle()
r.get_dim()
print("Area of Rectangle is", r.area())

# this is for triangle

t = Triangle()
t.get_dim()
print("Area of triangle is", t.area())

````

### `Problem-5:` Write a program with class Bill. The users have the option to pay the bill either by cheque or by cash. Use the inheritance to model this situation.

````python
class Bill:
    def __init__(self,amount):
        self.amount = amount

    def pay():
        pass


class Cashpayment(Bill):
    def pay(self):
        return f"pay rupee {self.amount} by cash"

class Chequepayment(Bill):
    def __init__(self, amount,chequeno):
        super().__init__(amount)
        self.chequeno = chequeno

    def pay(self):
        return f"pay rupee {self.amount} by cheque No: {self.chequeno}"


obj = Cashpayment(500)
obj1 = Chequepayment(400, 'CB543@')

obj1.pay()
````
