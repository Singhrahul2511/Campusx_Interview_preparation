# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

### Difference between Lists and Tuples

- Syntax
- Mutability
- Speed --> tuple fast hota hai kuki bhai fix size so fewer operstion iiliye operation jldi-jldi hoga 
- Memory  --> m/y km leta hai as compare to list
- Built in functionality --> list me jayda hota hai
- Error prone 
- Usability --> jaha bhot read krna hai data to to tuple use hoga aur jaha bhot sare changes perform krna hai waha List use hoga


```python
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
```

```python
import sys

L = list(range(1000))
T = tuple(range(1000))  # bhai tuple size bhi km occupy krta hai list se

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

## Error prone 

```python
a = [1,2,3]
b = a
             # lekin tuple me ye nhi hoga
a.append(4)
print(a)
print(b)  # example of shallow-copy (both variable a,b will point to the same obj so ak me change dusre me reflact hoga)
```

## Deep Copy

```python
import copy

a = [1, 2, 3]
b = copy.deepcopy(a) # Create a completely new, independent list
a.append(4)

print(f"a: {a}") # Output: a: [1, 2, 3, 4]
print(f"b: {b}") # Output: b: [1, 2, 3] (b remains unchanged)
```

```python
# tuple me variable assignment deep copy ko represent krta hai

a = (1,2,3)
b = a
  # tuple me ak me change kro to dusra me nhi hota hai bhai isiliye acha hai
a = a + (4,)
print(a)
print(b) # deep-copy ho jata hai(both a,b denotes diff obj)
```

# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

### Creating Sets

```python
# empty
s = set() #agr {} ye bnao tb dict bn jayega kuki dict and set in python same syntax share krte hai to default wala dict ko de diya gya hai.
# age aapko bnana hai to type-castinng-wala concept ka use kro like s = set()
print(s)
print(type(s))
# 1D and 2D
s1 = {1,2,3}
print(s1)
#s2 = {1,2,3,{4,5}} # ye error dega kuki set ke andar mutable data-type nhi rkh skte hai.lekin set khud mutable hai apne-app me
#print(s2)
# homo and hetro
s3 = {1,'hello',4.5,(1,2,3)}
print(s3)

# using type casting
s4 = set([1,2,3])
print(s4)
# duplicates not allowed
s5 = {1,1,2,2,3,3}
print(s5)
# set can't have mutable items
s6 = {1,2,[3,4]} # error kuki set me andar mutable value nhi hoga kuki List mutable hai
print(s6)

## WHILE printing set item order ku change ho ja rha hai kuki internally set hashing use krta hai wo decide krta hai position
```

```python
s5 = {1,1,2,2,3,3}
print(s5) # means agr duplicate hoga to usko remove krke print krega
```

```python
s4 = set([1,2,3])
# s4 = {[1, 2, 3]} # ye wala error dega--> kuki set ke andar mutable data-type hai store
print(s4)
```

### Set Functions

```python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse=True) # ye set nhi balki List return kr-ke dega
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8,9}

# s1 | s2
print(s1.union(s2))


s1.update(s2) # s1 ka value to update ke baad change ho gya
print(s1)
print(s2)
```

```python
# intersection/intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)
# s1.intersection_update(s2)
# print(s1)
print(s2)
```

```python
# difference/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

print(s1.difference(s2))

print(s1.difference_update(s2))


# print(s1)
# print(s2)
```

```python
# symmetric_difference/symmetric_difference_update --> same element remove ho jayega dono sets se
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint/issubset/issuperset --> if sare items alg hai to True nhi to False
s1 = {1,2,3,4}
s2 = {7,8,5,6}

s1.isdisjoint(s2)
```

```python
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2) # to bhai s1 to hai hi superset of s2 kuki sare element s2 ke s1 me hai
```

```python
# copy
s1 = {1,2,3}
s2 = s1.copy() # s2 bhi ak copy ho gya s1 ka element to same hoga lekin dono alg address pr store honge
s2.update([4,5])
print(s1)
print(s2) # yaha bhi Deep-Copy ho rha hai
```

```python
# copy
import copy
s1 = {1,2,3}
s2 = copy.deepcopy(s1) # s2 bhi ak copy ho gya s1 ka element to same hoga lekin dono alg address pr store honge
s1.update([4,5])

print(s1)
print(s2) # yaha bhi Deep-Copy ho rha hai
```

# Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

Characterstics:

- Mutable
- Indexing has no meaning --> key,val pare ko aap nhi nikal skte ho
- keys can't be duplicated
- keys can't be mutable items --> keys ko aap change nhi kr skte hai 

### Create Dictionary

```python
# empty dictionary
d = {}
d
# 1D dictionary
d1 = { 'name' : 'nitish' ,'gender' : 'male' }
d1
# with mixed keys
d2 = {(1,2,3):1,'hello':'world'} # yaha tuple aur string hai kuki both are immutable. agr mutable rhta to nhi ho pata
d2
# 2D dictionary -> JSON
s = {
    'name':'nitish',
     'college':'bit',
     'sem':4,
     'subjects':{
         'dsa':50,
         'maths':67,
         'english':34
     }
}
s
# using sequence and dict function
d4 = dict([('name','nitish'),('age',32),(3,3)])
print(d4)
# duplicate keys
d5 = {'name':'nitish','name':'rahul'}
d5
# mutable items as keys
d6 = {'name':'nitish',(1,2,3):2}
print(d6)

# dict are unordered
s['subjects']['maths']
```
