###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
# write your code here
tup = (1,5,7,8,10)
lst= []
lst.append(tup[0]*tup[1])
for i in range(1,len(tup)-1):
    lst.append(tup[i-1]*tup[i] + tup[i]*tup[i+1])
lst.append(tup[-1]*tup[-2])
print(lst)
````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
# write your code here
t1 = (1,2,3,0)
t2 = (0,1,2,3)
flag = True
for i,j in zip(t1,t2):
    if i==j:
        continue
    else:
        flag = False
        break
if flag:
    print("same")
else:
    print("not same")

# not same kuki (list,tuple) me order matter krta hai sare element same hai but diff-diff position pr.isiliye dono ka element same position pr rha chaiye
````

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
# write your code here
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]
newlst = [0]*len(list1)
for i in list1:
    if type(i) == list:
        newlst[0] +=1
    elif type(i) == set:
        newlst[1] +=1
    elif type(i) == tuple:
        newlst[2] +=1
    else:
        pass
print("List-{}\nSet-{}\nTuple-{}".format(newlst[0],newlst[1],newlst[2]))
````

### `Q1`: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

Input:

```bash
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
CampusX
```

**Example 2:**

Input:
```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
Best
```

````python
# write your code here
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
max_len = 0
max_key = ""
for i in test_dict:
    if max_len < len(set(test_dict[i])):
        max_len = len(set(test_dict[i]))
        max_key = i
print(max_key)
````


Output:

```bash
    'CampusX is the best channel for Data-Science students.
```

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.
```
