### 2.Find how many movies have a rating > the avg of all the movie ratings(Find the count of above average movies)


SELECT count(*) FROM campusx.movies
where score > (select avg(score) from campusx.movies)



### 3. Find the highest rated movie of 2000.


select * from campusx.movies
where year = 2000 and score = (select max(score) from campusx.movies
where year = 2000)



### 4. Find the highest rated movie among all movies whose number of votes are > the dataset avg votes

select * from campusx.movies
where score = (SELECT max(score) FROM campusx.movies

# ROW- SUB QUESRY

- ROW SUB-QERIES = jo multiple row return krta hai

 - `1. Find all users who never ordered1.`


select * from zomato.users
where user_id not in (SELECT distinct(user_id) FROM zomato.orders) 

- isko row-query ku bol rhe hai kuki ye jb inner-query is running then it is giving multiple row f user_id isiliye



### 2. Find all the movies made by top 3 directors(in terms of total gross income)

select * from campusx.movies
where director in (SELECT director FROM campusx.movies
group by director
order by sum(gross) desc
)



with top_director as (SELECT director FROM campusx.movies
group by director
order by sum(gross) desc limit 5
)
select * from campusx.movies
where director in (select * from top_director)





### 3. Find all movies of all those actors whose filmography's avg rating > 8.5(take 25000 votes as cutoff)


select * from campusx.movies
where star in ( SELECT star FROM campusx.movies
					where votes>25000
					group by star
					having avg(score)>8.5)

# Independent-subquery (Table Subquery(multiple column and multiple rows))

- `1. FInd the most profitable movie of each year`

select * from campusx.movies
where (year,gross-budget) in (SELECT year,max(gross-budget) FROM campusx.movies
group by year)






- `note: scalar wale me (=) lgate the ku ki single value return krta tha lekin multiple value return krne keliye (IN or NOT-IN) use krna hai`

- `3. find the highest grossing movies of top 5 actor/director combo in terms of total gross income`


use campusx;

with top_duos as (
				select star,director,max(gross) from movies
				group by star,director
				order by sum(gross) desc limit 5
                )

select * from movies 
where (star,director,gross) in (select* from top_duos)
